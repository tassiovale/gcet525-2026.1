# Testes do Parser (sem arquivo)

Notebook para validar sintaxe e visualizar AST diretamente a partir de strings.

```python
import json

from parser import ParserError, parse_source


def testar(codigo: str, mostrar_ast: bool = True):
    try:
        ast = parse_source(codigo)
        print("Sintaxe OK")

        if mostrar_ast:
            payload = ast.to_dict() if hasattr(ast, "to_dict") else ast
            print(json.dumps(payload, indent=2, ensure_ascii=False))

        return ast
    except ParserError as exc:
        print(f"Erro sintático: {exc}")
        return None
```

```python
codigo_ok = """
main {
  int x = 10;
  int y = 5;
  bool ok = true;

  x = x + 2 * (3 + y);

  if (x > y && ok) {
    print(x, y, ok);
  } else {
    read(x);
  }

  while (x > 0) {
    x = x - 1;
  }
}
"""

testar(codigo_ok)
```

```python
codigo_erro = """
main {
  int x = ;
}
"""

testar(codigo_erro, mostrar_ast=False)
```
