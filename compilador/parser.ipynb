# Parser (Descida Recursiva)

Notebook para testar o parser diretamente com strings, sem precisar de arquivo de entrada.

```python
import json

from parser import ParserError, parse_source


def testar(codigo: str, mostrar_ast: bool = True):
    try:
        ast = parse_source(codigo)
        print("Sintaxe OK")
        if mostrar_ast:
            print(json.dumps(ast, indent=2, ensure_ascii=False))
        return ast
    except ParserError as exc:
        print(f"Erro sintático: {exc}")
        return None
```

```python
codigo_ok = """
main {
  int x = 10;
  bool ok = true;
  x = x + 2 * (3 + 1);
  if (x > 0 && ok) {
    print(x, ok);
  } else {
    read(x);
  }
  while (x > 0) {
    x = x - 1;
  }
}
"""

codigo_ok = """
main {
  int x = 10;
  int y = 5;
  if(x>y){
      print(x);
  }
}
"""

testar(codigo_ok)
```
